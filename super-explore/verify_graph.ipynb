# Verify Graph Capabilities
This notebook verifies the GraphManager implementation and visualizes the Super Power graph.

```python
import os
from IPython.display import display, HTML
from super.core.graph import GraphManager
```

```python
# Initialize GraphManager (loads warehouse data)
# Assuming we are in notebooks/ directory, warehouse is at ../../data/warehouse
# If running via package, maybe default logic works, but let's be explicit for demo
warehouse_path = os.path.abspath('../../data/warehouse')
    
print(f"Loading warehouse from: {warehouse_path}")
gm = GraphManager(warehouse_root=warehouse_path)
print(f"Graph Created! Nodes: {gm.G.number_of_nodes()}, Edges: {gm.G.number_of_edges()}")
```

```python
# Query Flying Heroes
flying_heroes = gm.get_flying_heroes()
print(f"Flying Heroes found: {len(flying_heroes)}")
for h in flying_heroes:
    print(f"- {h}")
```

```python
# Visualize Subgraph for first flying hero
if flying_heroes:
    hero = flying_heroes[0]
    print(f"Visualizing subgraph for {hero}...")
    sub_G = gm.subgraph_for_hero(hero, depth=2)
    if sub_G:
        # Use filename=None to get HTML string
        html = gm.visualize(sub_G, filename=None)
        display(HTML(html))
```
